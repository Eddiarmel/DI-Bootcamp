```python
# ==========================================================
# PROJET : ANALYSE DU DATASET US SUPERSTORE
# ==========================================================

# 1. IMPORTATION DES BIBLIOTHEQUES
# ==========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use('ggplot')

# ==========================================================
# 2. CHARGEMENT DES DONNEES
# ==========================================================

df = pd.read_excel('US Superstore data.xls')

print("Dimensions du dataset :", df.shape)

print("\nValeurs manquantes :")
print(df.isnull().sum())

print("\nDoublons :", df.duplicated().sum())

# Suppression des doublons
df = df.drop_duplicates()

# Conversion des dates
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])

print("\nAperçu du dataset")
print(df.head())

# ==========================================================
# QUESTION 1 :
# Quels sont les États qui enregistrent le plus de ventes ?
# ==========================================================

state_sales = (
    df.groupby("State")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTOP ETATS PAR VENTES")
print(state_sales.head(10))

plt.figure(figsize=(12,6))
state_sales.head(10).plot(kind='bar')
plt.title("Top 10 des États par ventes")
plt.ylabel("Ventes")
plt.show()

# ==========================================================
# QUESTION 2 :
# Comparaison New York vs Californie
# ==========================================================

ny_ca = df[df["State"].isin(["New York","California"])]

comparison = ny_ca.groupby("State")[["Sales","Profit"]].sum()

print("\nNEW YORK VS CALIFORNIA")
print(comparison)

comparison.plot(kind='bar', figsize=(8,5))
plt.title("Comparaison CA et Profit")
plt.show()

# ==========================================================
# QUESTION 3 :
# Client exceptionnel à New York
# ==========================================================

ny = df[df["State"]=="New York"]

top_customer_ny = (
    ny.groupby("Customer Name")
    .agg({
        "Sales":"sum",
        "Profit":"sum"
    })
    .sort_values("Sales", ascending=False)
)

print("\nTOP CLIENTS NEW YORK")
print(top_customer_ny.head(10))

# ==========================================================
# QUESTION 4 :
# Différences de rentabilité entre les États
# ==========================================================

state_profit = (
    df.groupby("State")
    .agg({
        "Sales":"sum",
        "Profit":"sum"
    })
)

state_profit["Profitability"] = (
    state_profit["Profit"] /
    state_profit["Sales"] * 100
)

state_profit = state_profit.sort_values(
    "Profitability",
    ascending=False
)

print("\nETATS LES PLUS RENTABLES")
print(state_profit.head(10))

print("\nETATS LES MOINS RENTABLES")
print(state_profit.tail(10))

plt.figure(figsize=(14,6))
state_profit["Profitability"].head(15).plot(kind="bar")
plt.title("Top États les plus rentables")
plt.ylabel("% Rentabilité")
plt.show()

# ==========================================================
# QUESTION 5 :
# Principe de Pareto (clients et bénéfices)
# ==========================================================

customer_profit = (
    df.groupby("Customer Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

cum_profit = customer_profit.cumsum()

cum_profit_pct = (
    cum_profit /
    customer_profit.sum()
)

customer_pct = (
    np.arange(1, len(customer_profit)+1)
    / len(customer_profit)
)

plt.figure(figsize=(10,6))
plt.plot(customer_pct*100, cum_profit_pct*100)
plt.axvline(20, linestyle="--")
plt.axhline(80, linestyle="--")
plt.title("Pareto : Clients vs Bénéfices")
plt.xlabel("% Clients")
plt.ylabel("% Bénéfices cumulés")
plt.grid()
plt.show()

n20 = int(len(customer_profit)*0.2)

profit_20 = (
    customer_profit.head(n20).sum()
    / customer_profit.sum()
    *100
)

print(
    f"\nLes 20% des clients génèrent {profit_20:.2f}% des bénéfices"
)

# ==========================================================
# QUESTION 6 :
# Top 20 villes par chiffre d'affaires
# ==========================================================

city_sales = (
    df.groupby("City")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(20)
)

print("\nTOP 20 VILLES PAR VENTES")
print(city_sales)

plt.figure(figsize=(14,6))
city_sales.plot(kind='bar')
plt.title("Top 20 villes par ventes")
plt.ylabel("Ventes")
plt.show()

# ==========================================================
# QUESTION 7 :
# Top 20 villes par bénéfice
# ==========================================================

city_profit = (
    df.groupby("City")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(20)
)

print("\nTOP 20 VILLES PAR PROFIT")
print(city_profit)

plt.figure(figsize=(14,6))
city_profit.plot(kind='bar')
plt.title("Top 20 villes par bénéfice")
plt.ylabel("Profit")
plt.show()

# ==========================================================
# QUESTION 8 :
# Rentabilité des villes
# ==========================================================

city_stats = (
    df.groupby("City")
    .agg({
        "Sales":"sum",
        "Profit":"sum"
    })
)

city_stats["Profitability"] = (
    city_stats["Profit"]
    / city_stats["Sales"] * 100
)

city_stats = city_stats.sort_values(
    "Profitability",
    ascending=False
)

print("\nVILLES LES PLUS RENTABLES")
print(city_stats.head(10))

print("\nVILLES LES MOINS RENTABLES")
print(city_stats.tail(10))

# ==========================================================
# QUESTION 9 :
# Top 20 clients par ventes
# ==========================================================

top_customers = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(20)
)

print("\nTOP 20 CLIENTS PAR VENTES")
print(top_customers)

plt.figure(figsize=(14,6))
top_customers.plot(kind='bar')
plt.title("Top 20 clients par ventes")
plt.ylabel("Ventes")
plt.show()

# ==========================================================
# QUESTION 10 :
# Courbe cumulative des ventes
# ==========================================================

customer_sales = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

cum_sales = customer_sales.cumsum()

cum_sales_pct = (
    cum_sales
    / customer_sales.sum()
)

customer_pct_sales = (
    np.arange(1, len(customer_sales)+1)
    / len(customer_sales)
)

plt.figure(figsize=(10,6))
plt.plot(
    customer_pct_sales*100,
    cum_sales_pct*100
)

plt.axvline(20, linestyle="--")
plt.axhline(80, linestyle="--")

plt.title("Pareto : Clients vs Ventes")
plt.xlabel("% Clients")
plt.ylabel("% Ventes cumulées")
plt.grid()
plt.show()

n20_sales = int(len(customer_sales)*0.2)

sales_20 = (
    customer_sales.head(n20_sales).sum()
    / customer_sales.sum()
    *100
)

print(
    f"\nLes 20% des clients génèrent {sales_20:.2f}% des ventes"
)

# ==========================================================
# QUESTION 11 :
# Recommandations marketing
# ==========================================================

print("\n==============================")
print("RECOMMANDATIONS MARKETING")
print("==============================")

best_states = (
    df.groupby("State")
    .agg({
        "Sales":"sum",
        "Profit":"sum"
    })
    .sort_values(
        ["Profit","Sales"],
        ascending=False
    )
)

print("\nÉtats à privilégier :")
print(best_states.head(10))

best_cities = (
    df.groupby("City")
    .agg({
        "Sales":"sum",
        "Profit":"sum"
    })
    .sort_values(
        ["Profit","Sales"],
        ascending=False
    )
)

print("\nVilles à privilégier :")
print(best_cities.head(10))

print("\nAnalyse terminée avec succès.")
```
