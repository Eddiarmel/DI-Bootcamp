```python

# Créez une matrice 3x3 et effectuez les opérations suivantes
# 1. Créez une matrice 3x3 avec des valeurs de 1 à 9
matrice = np.array([[1, 9, 3],
                    [4, 5, 6],
                    [7, 8, 9]])
print("Matrice 3x3:")
print(matrice)
#Calculez le déterminant.
determinant = np.linalg.det(matrice)
print("\nDéterminant de la matrice:")
print(determinant)
# 2. Calculez l'inverse de la matrice (si elle est inversible).
if determinant != 0:
    inverse = np.linalg.inv(matrice)
    print("\nInverse de la matrice:")
    print(inverse)
```

```python
import numpy as np

np.random.seed(42)

# Données
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after  = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Hypothèses
# H0 : la formation n'a PAS d'effet  → moyenne des différences = 0
# H1 : la formation améliore la productivité → moyenne des diff > 0

differences = productivity_after - productivity_before
n = len(differences)

moyenne_diff  = np.mean(differences)
ecart_type_diff = np.std(differences, ddof=1)  # ddof=1 → écart-type échantillon

# Statistique t appairée
t_stat = moyenne_diff / (ecart_type_diff / np.sqrt(n))

print(f"Moyenne avant  : {np.mean(productivity_before):.2f}")
print(f"Moyenne après  : {np.mean(productivity_after):.2f}")
print(f"Moyenne diff   : {moyenne_diff:.2f}")
print(f"Écart-type diff: {ecart_type_diff:.2f}")
print(f"Statistique t  : {t_stat:.4f}")
print()

# Règle de décision (seuil α = 0.05, valeur critique ≈ 1.699 pour n=30)
seuil_critique = 1.699
if t_stat > seuil_critique:
    print("✅ Conclusion : On REJETTE H0.")
    print("   La formation a amélioré la productivité de façon significative.")
else:
    print("❌ Conclusion : On NE rejette PAS H0.")
    print("   Pas assez de preuves d'une amélioration significative.")
```

```python
import numpy as np
import pandas as pd

np.random.seed(0)

# Générer des données quotidiennes pour 2023
dates = pd.date_range(start='2023-01-01', end='2023-12-31', freq='D')
valeurs = np.random.randint(100, 500, size=len(dates))

serie = pd.Series(valeurs, index=dates)
print(f"Série complète : {len(serie)} jours\n")

# Découpage par trimestre
T1 = serie['2023-01':'2023-03'] 
T2 = serie['2023-04':'2023-06']  
T3 = serie['2023-07':'2023-09']  
T4 = serie['2023-10':'2023-12']  

for nom, trimestre in [('T1 (Jan–Mar)', T1), ('T2 (Avr–Jun)', T2),
                        ('T3 (Jul–Sep)', T3), ('T4 (Oct–Déc)', T4)]:
    print(f"{nom} → {len(trimestre)} jours | "
          f"Moyenne: {trimestre.mean():.1f} | "
          f"Min: {trimestre.min()} | Max: {trimestre.max()}")
```

🌟 Exercice 10 : Visualisation de base

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(7)

# Générer les données
x = np.arange(1, 51)                      # indices 1 à 50
y = np.random.randint(10, 100, size=50)   # valeurs aléatoires

# Calcul de la moyenne mobile (fenêtre de 5)
moyenne_mobile = np.convolve(y, np.ones(5)/5, mode='valid')

# ─── Création du graphique ─────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 4))

ax.plot(x, y, color='steelblue', linewidth=1.2,
        marker='o', markersize=4, label='Valeurs aléatoires')

ax.plot(x[2:-2], moyenne_mobile, color='tomato', linewidth=2,
        linestyle='--', label='Moyenne mobile (fenêtre=5)')

ax.axhline(y=np.mean(y), color='gray', linestyle=':', linewidth=1.2,
           label=f'Moyenne globale ({np.mean(y):.1f})')

ax.set_title("Graphique linéaire — 50 nombres aléatoires", fontsize=13)
ax.set_xlabel("Index")
ax.set_ylabel("Valeur")
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
