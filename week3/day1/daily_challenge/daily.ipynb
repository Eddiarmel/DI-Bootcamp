```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns

np.random.seed(42)  

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

cities = ['Paris', 'Tokyo', 'Cairo', 'Oslo', 'Sydney',
          'Mumbai', 'New York', 'Cape Town', 'Beijing', 'São Paulo']

# Génération des températures synthétiques : 10 villes × 12 mois
synthetic_data = np.random.uniform(-5, 35, (10, 12))

# Création du DataFrame
df = pd.DataFrame(synthetic_data, index=cities, columns=months)

print('✅ DataFrame créé avec succès')
print(f'   Dimensions : {df.shape[0]} villes × {df.shape[1]} mois')
df.round(1)
```

2 analyse

```python
# Température moyenne annuelle par ville
annual_mean = df.mean(axis=1).round(2)

# Ville la plus chaude et la plus froide
hottest_city = annual_mean.idxmax()
coldest_city = annual_mean.idxmin()

print('  Température moyenne annuelle par ville :')
print(annual_mean.to_string())
print(f' Ville la plus chaude : {hottest_city} ({annual_mean[hottest_city]:.2f}°C)')
print(f' Ville la plus froide : {coldest_city} ({annual_mean[coldest_city]:.2f}°C)')
```

3 visualisation

```python
import numpy as np
import matplotlib.pyplot as plt

# Mois
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

# Données aléatoires
np.random.seed(42)

# Couleur verte principale
green = '#1b9e77'

plt.figure(figsize=(12, 6))

for i in range(10):
    temperatures = np.random.uniform(-5, 35, 12)

    # Villes 7 et 10 en pointillés
    if i in [6, 9]:
        plt.plot(
            months,
            temperatures,
            color=green,
            linestyle='--',
            linewidth=3.5,
            alpha=1,
            label=f'City {i+1}'
        )
    else:
        plt.plot(
            months,
            temperatures,
            color=green,
            linewidth=2,
            alpha=0.7,
            label=f'City {i+1}'
        )

plt.title('Monthly Temperature Trends for Each City', fontsize=20)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Temperature (°C)', fontsize=14)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='lower left')

plt.tight_layout()
plt.show()
```
