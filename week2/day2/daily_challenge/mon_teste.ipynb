```python
import pandas as pd

data = {
    'Book Title': ['The Great Gatsby', 'To Kill a Mockingbird', '1984', 'Pride and Prejudice', 'The Catcher in the Rye'],
    'Author': ['F. Scott Fitzgerald', 'Harper Lee', 'George Orwell', 'Jane Austen', 'J.D. Salinger'],
    'Genre': ['Classic', 'Classic', 'Dystopian', 'Classic', 'Classic'],
    'Price': [10.99, 8.99, 7.99, 11.99, 9.99],
    'Copies Sold': [500, 600, 800, 300, 450]
}

df = pd.DataFrame(data)
```

```python
# affichons le 5 premiers lignes du dataframe
print(df.head())
```

```python
## faire le resumer statistique du dataframe
print(df.describe())

print("")
# pour avoir de information sur le dataframe
print(df.info())
```

```python
# pour trie par prix 
le_prix=df.sort_values(by='Price')
print(le_prix)
```

```python
## realiser un triage par genre
print(df[df['Genre'] == 'Classic'])
```

```python

# 7. Total ventes par auteur
print("\n=== Ventes par auteur ===")
print(df.groupby('Author')['Copies Sold'].sum())
```

```python
import pandas as pd

data = {
    'Book Title': ['The Great Gatsby', 'To Kill a Mockingbird', '1984', 'Pride and Prejudice', 'The Catcher in the Rye'],
    'Author': ['F. Scott Fitzgerald', 'Harper Lee', 'George Orwell', 'Jane Austen', 'J.D. Salinger'],
    'Genre': ['Classic', 'Classic', 'Dystopian', 'Classic', 'Classic'],
    'Price': [10.99, 8.99, 7.99, 11.99, 9.99],
    'Copies Sold': [500, 600, 800, 300, 450]
}

eddi = pd.DataFrame(data)
## cette commande permet d afficher les 5 premiers lignes du dataframe
print(eddi.head())
### avec la commande describe on peut faire un resumer statistique du dataframe
print(eddi.describe())

# pour obtenir un resumer de l information sur le dataframe
print(eddi.info())

# Triez le DataFrame en fonction du prix ou du nombre d'exemplaires vendus.
print(eddi.sort_values(by='Price'))
print(eddi.sort_values(by='Copies Sold'))

# Filtrer les livres par genre spécifique ou par prix supérieur à un certain seuil.
print(eddi[eddi['Genre']=='Classic'])
# Classez les livres par auteur et additionnez le nombre d'exemplaires vendus
print(eddi.groupby('Author')['Copies Sold'].sum())
```
